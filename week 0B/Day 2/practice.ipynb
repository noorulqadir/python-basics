Rectangle class. Attributes: width and height. Methods: area() and perimeter(), and a __str__ that shows Rectangle 4 x 5.
Test: Rectangle(4, 5).area() gives 20, and perimeter() gives 18.

```python
class Rectangle:
    def __init__(self, width, height):
        self.width = width
        self.height = height

    def area(self):
        return self.width * self.height

    def perimeter(self):
        return 2 * (self.width + self.height)

    def __str__(self):
        return f"Rectangle {self.width} x {self.height}"

r = Rectangle(4, 5)
print(r)
print("Area: ", r.area())
print("Perimeter: ", r.perimeter())
```

Expense class (the tracker preview). Attributes: amount, category, description. Add __str__ to show 12.50 | food | lunch. Then write a standalone function total_amount(expenses) that takes a list of Expense objects and returns the sum.
Test: make 3 objects, put them in a list, and print the total.

```python
class Expense:
    def __init__(self, amount, category, description):
        self.amount = amount
        self.category = category
        self.description = description

    def __str__(self):
        return f"{self.amount:.2f} | {self.category} | {self.description}"

def total_amount(expenses):
    total = 0 
    for expense in expenses:
        total = total + expense.amount
    return total

e1 = Expense(12.50, "food", "lunch")
e2 = Expense(20.00, "travel", "bus")
e3 = Expense(15.50, "shopping", "books")

expenses = [e1, e2, e3]

for expense in expenses:
    print(expense)

print("Total Amount: ", total_amount(expenses))
```

BankAccount class with checks. Extend the example above so withdraw and deposit reject negative amounts, and add a history list attribute that records each transaction as a string (for example, "deposit 50"). Add a method show_history() that prints them.
Hint: create the list in __init__ with self.history = [], and use .append() in each method.

```python
class BankAccount:
    def __init__(self, balance = 0):
        self.balance = balance
        self.history = []

    def deposit(self, amount):
        if amount <= 0:
            print("Error: Deposit amount must be positive")
        else:
            self.balance = self.balance + amount
            self.history.append(f"deposit {amount}")
            print("Deposit successful")

    def withdraw(self, amount):
        if amount <= 0:
            print("Error: withdraw amount must be positve")
        elif amount > self.balance:
            print("Insufficient balance")
        else:
            self.balance = self.balance - amount
            self.history.append(f"withdraw {amount}")
            print("Withdraw successful")

    def show_balance(self):
        print("Current balance: ", self.balance)

    def show_history(self):
        print("Transaction history: ")
        for transaction in self.history:
            print(transaction)

account = BankAccount(100)
account.deposit(50)
account.withdraw(30)
account.deposit(200)
account.withdraw(-10)
account.deposit(-20)

account.show_balance()
account.show_history()
```
