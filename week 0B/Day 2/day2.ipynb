# Classes

So far, data (a dictionary) and behavior (functions) have been separate. A class bundles them together into one thing. Compare:

```python
# without class
expense = {"amount": 12.5, "category": "food"}
def describe(expense):
    return f"{expense["category"]}: {expense["amount"]}"
```

```python
# with class
class Expense:
    def __init__(self, amount, category):
        self.amount = amount
        self.category = category

    def describe(self):
        return f"{self.category}: {self.amount}"
```

```python
lunch = Expense(12.5, "food")
taxi = Expense(40.0, "transport")

print(lunch.amount)
print(taxi.category)
print(lunch.describe)
```

# Method that change data

```python
class BankAccount:
    def __init__(self, owner, balance = 0):
        self.owner = owner
        self.balance = balance

    def deposit(self, amount):
        self.balance = self.balance + amount

    def withdraw(self, amount):
        if amount > self.balance:
            print("Insufficiant balance")
        else:
            self.balance = self.balance - amount

account = BankAccount("sara", 100)
account.deposit(50)
account.withdraw(30)
print(account.balance)
```

# Making objects printable with __str__

```python
class Expense:
    def __init__(self, amount, category):
        self.amount = amount
        self.category = category

    def __str__(self):
        return f"{self.category}: {self.amount:.2f}"

lunch = Expense(12.5, "food")
print(lunch)
```

Without __str__, printing shows an ugly <__main__.Expense object at 0x...>. Names with double underscores on both sides are called special methods, and Python calls them automatically.

# Lists of objects

```python
expenses = [Expense(12.5, "food"), Expense(40, "transport"), Expense(8, "food")]

total = 0
for e in expenses:
    total = total + e.amount
print(total)
```

```python
class counter:
    def __init__(self):
        self.count = 0

    def add(self):
        self.count = self.count + 1

    def reset(self):
        self.count = 0

    def __str__(self):
        return f"count: {self.count}"

c = counter()
c.add()
c.add()
print(c)
c.reset()
print(c)
```
